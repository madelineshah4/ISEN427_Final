```python
import pymc as pm
import numpy as np
import pandas as pd
import arviz as az

# Load data
data = pd.read_csv('cleaned/Umpire_1_cleaned.csv')

features = [
    'balls', 'strikes', 'zone', 'pfx_x', 'pfx_z', 'plate_x', 'plate_z', 'error_in_decision',
    'pitch_location', 'delta_home_win_exp', 'delta_run_exp', 'vx0', 'vy0', 'vz0', 'ay',
    'sz_top', 'sz_bot', 'home_score', 'away_score', 'all_star_player',
    'pitch_type_CH', 'pitch_type_CS', 'pitch_type_CU', 'pitch_type_EP', 'pitch_type_FA',
    'pitch_type_FC', 'pitch_type_FF', 'pitch_type_FO', 'pitch_type_FS', 'pitch_type_KC',
    'pitch_type_KN', 'pitch_type_SC', 'pitch_type_SI', 'pitch_type_SL', 'pitch_type_ST',
    'pitch_type_SV', 'description_called_strike', 'stand_R', 'p_throws_R'
]

coords = {'n_obs': range(len(data))}

# Priors
with pm.Model(coords=coords) as model:
    # Intercept
    intercept = pm.Normal('intercept', mu=0, sigma=2)

    # Coefficients
    beta_balls = pm.Normal('beta_balls', mu=0, sigma=2)
    beta_strikes = pm.Normal('beta_strikes', mu=0, sigma=2)
    beta_zone = pm.Normal('beta_zone', mu=1, sigma=2)
    beta_pfx_x = pm.Normal('beta_pfx_x', mu=1, sigma=2)
    beta_pfx_z = pm.Normal('beta_pfx_z', mu=0, sigma=2)
    beta_plate_x = pm.Normal('beta_plate_x', mu=-1, sigma=2)
    beta_plate_z = pm.Normal('beta_plate_z', mu=-1, sigma=2)
    beta_error_in_decision = pm.Normal('beta_error_in_decision', mu=0, sigma=2)
    beta_pitch_location = pm.Normal('beta_pitch_location', mu=-2, sigma=2)
    beta_delta_home_win_exp = pm.Normal('beta_delta_home_win_exp', mu=0, sigma=2)
    beta_delta_run_exp = pm.Normal('beta_delta_run_exp', mu=-2, sigma=2)
    beta_vx0 = pm.Normal('beta_vx0', mu=0, sigma=2)
    beta_vy0 = pm.Normal('beta_vy0', mu=0, sigma=2)
    beta_vz0 = pm.Normal('beta_vz0', mu=0, sigma=2)
    beta_ay = pm.Normal('beta_ay', mu=0, sigma=2)
    beta_sz_top = pm.Normal('beta_sz_top', mu=1, sigma=2)
    beta_sz_bot = pm.Normal('beta_sz_bot', mu=1, sigma=2)
    beta_home_score = pm.Normal('beta_home_score', mu=0, sigma=2)
    beta_away_score = pm.Normal('beta_away_score', mu=0, sigma=2)
    beta_all_star_player = pm.Normal('beta_all_star_player', mu=0, sigma=2)
    beta_pitch_type_CH = pm.Normal('beta_pitch_type_CH', mu=1, sigma=2)
    beta_pitch_type_CS = pm.Normal('beta_pitch_type_CS', mu=0, sigma=2)
    beta_pitch_type_CU = pm.Normal('beta_pitch_type_CU', mu=1, sigma=2)
    beta_pitch_type_EP = pm.Normal('beta_pitch_type_EP', mu=-1, sigma=2)
    beta_pitch_type_FA = pm.Normal('beta_pitch_type_FA', mu=1, sigma=2)
    beta_pitch_type_FC = pm.Normal('beta_pitch_type_FC', mu=-0.5, sigma=2)
    beta_pitch_type_FF = pm.Normal('beta_pitch_type_FF', mu=1, sigma=2)
    beta_pitch_type_FO = pm.Normal('beta_pitch_type_FO', mu=-1, sigma=2)
    beta_pitch_type_FS = pm.Normal('beta_pitch_type_FS', mu=1, sigma=2)
    beta_pitch_type_KC = pm.Normal('beta_pitch_type_KC', mu=-1, sigma=2)
    beta_pitch_type_KN = pm.Normal('beta_pitch_type_KN', mu=-1, sigma=2)
    beta_pitch_type_SC = pm.Normal('beta_pitch_type_SC', mu=-1, sigma=2)
    beta_pitch_type_SI = pm.Normal('beta_pitch_type_SI', mu=-1, sigma=2)
    beta_pitch_type_SL = pm.Normal('beta_pitch_type_SL', mu=-1, sigma=2)
    beta_pitch_type_ST = pm.Normal('beta_pitch_type_ST', mu=1, sigma=2)
    beta_pitch_type_SV = pm.Normal('beta_pitch_type_SV', mu=-1, sigma=2)
    beta_description_called_strike = pm.Normal('beta_description_called_strike', mu=0, sigma=2)
    beta_stand_R = pm.Normal('beta_stand_R', mu=0, sigma=2)
    beta_p_throws_R = pm.Normal('beta_p_throws_R', mu=0, sigma=2)

    # Logits
    logits = (
        intercept
        + beta_balls * data['balls']
        + beta_strikes * data['strikes']
        + beta_zone * data['zone']
        + beta_pfx_x * data['pfx_x']
        + beta_pfx_z * data['pfx_z']
        + beta_plate_x * data['plate_x']
        + beta_plate_z * data['plate_z']
        + beta_error_in_decision * data['error_in_decision']
        + beta_pitch_location * data['pitch_location']
        + beta_delta_home_win_exp * data['delta_home_win_exp']
        + beta_delta_run_exp * data['delta_run_exp']
        + beta_vx0 * data['vx0']
        + beta_vy0 * data['vy0']
        + beta_vz0 * data['vz0']
        + beta_ay * data['ay']
        + beta_sz_top * data['sz_top']
        + beta_sz_bot * data['sz_bot']
        + beta_home_score * data['home_score']
        + beta_away_score * data['away_score']
        + beta_all_star_player * data['all_star_player']
        + beta_pitch_type_CH * data['pitch_type_CH']
        + beta_pitch_type_CS * data['pitch_type_CS']
        + beta_pitch_type_CU * data['pitch_type_CU']
        + beta_pitch_type_EP * data['pitch_type_EP']
        + beta_pitch_type_FA * data['pitch_type_FA']
        + beta_pitch_type_FC * data['pitch_type_FC']
        + beta_pitch_type_FF * data['pitch_type_FF']
        + beta_pitch_type_FO * data['pitch_type_FO']
        + beta_pitch_type_FS * data['pitch_type_FS']
        + beta_pitch_type_KC * data['pitch_type_KC']
        + beta_pitch_type_KN * data['pitch_type_KN']
        + beta_pitch_type_SC * data['pitch_type_SC']
        + beta_pitch_type_SI * data['pitch_type_SI']
        + beta_pitch_type_SL * data['pitch_type_SL']
        + beta_pitch_type_ST * data['pitch_type_ST']
        + beta_pitch_type_SV * data['pitch_type_SV']
        + beta_description_called_strike * data['description_called_strike']
        + beta_stand_R * data['stand_R']
        + beta_p_throws_R * data['p_throws_R']
    )

    # Transform logits to probabilities using the sigmoid function
    p = pm.Deterministic('p', pm.math.sigmoid(logits))

    # Bernoulli likelihood
    y = pm.Bernoulli('y', p=p, observed=data['error_in_decision'])

    # Sample from prior predictive
    prior_pred = pm.sample_prior_predictive(samples=500)
```
